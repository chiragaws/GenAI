```python
import re
text = """
Artificial Intelligence is transforming the banking sector.
Banks use AI for fraud detection, customer support, and risk analysis.
Machine learning models help in predicting loan defaults.
Natural Language Processing is used in chatbots and document analysis.
Deep learning improves credit scoring accuracy.
"""
# Step 1: Split by sentences
chunks = re.split(r'(?<=[.!?])\s+', text.strip())
# for i, chunk in enumerate(chunks):
# print(f"Chunk {i+1}: {chunk}")
print(chunks)
```

```python
def get_embeddings(chunk):
  from openai import OpenAI
  client = OpenAI()
  response = client.embeddings.create(
      input = chunks[0] ,model="text-embedding-3-small"

  )
  return (response.data[0].embedding)
```

```python
len(response.data[0].embedding)
```

```python
vectors = []
for chunk in chunks:
  result = get_embeddings(chunk)
  vectors.append(result)

vectors
```
