```python
from langchain_tavily import TavilySearch
tool = TavilySearch(
    max_results=5,
    topic="general"
)
```

```python
#This is generated by a model, but we'll create a tool call directly
model_generated_tool_call ={
    "args" : {"query":"can you tell me T20I schedule 2026 for India vs WestIndies"},
    "id" : "1",
    "name" : "tavily",
    "type" : "tool_call",
}
tool_msg = tool.invoke(model_generated_tool_call)

#The content is a JSON string of results
print(tool_msg.content[:400])
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(
    model="gpt-4o"
)
```

```python
messages = [
    (
        "system",
        "You are a helpful assistant who understands Cricket and schedule. I want your help to generate the analysis in HTML along with schedule of India vs Westindies 2026 T20I schedule in tabular form"
    ),
    ("user",tool_msg.content),
]
#Generating the Results
ai_msg =llm.invoke(messages)
ai_msg.content
```

```python
from IPython.display import HTML,display
html_content = ai_msg.content
display (HTML(html_content))
```
