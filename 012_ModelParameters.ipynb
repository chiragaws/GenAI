```python
"""TOP_K(1-100+) - Hugging face models
- Filtering data generated from the temperature
- Purchasing vegetables by filtering from local vendor where that vendor will
alreay filter data while he is purchasing it from big vendor
When to use:
Low (1-10): When you want very focused, predictable responses
Medium (20-50): Balanced creativity and coherence
High (50-100): When you want more creative flexibility"""
```

```python
"""top_p ( Nucleus Sampling - 0.0 - 1.0)
top_p always take cumulative probability
When to use:
Low (0.1-0.3): Factual content, formal writing, technical explanations
Medium (0.4-0.7): Conversational responses, general content
High (0.8-0.95): Creative writing, brainstorming, varied responses"""
```

```python
"""Repetition Penality (0.8 - 1.5) - Hugging face models
Name it self caries a meaning that stop repeting
samething again and again. If it get repeated
then give the penality in terms of reducing probability score
TagLine: “Don’t keep saying the same thing again.”
When to use:
Low (0.9-1.0): When some repetition is acceptable (poetry, emphasis)
Medium (1.1-1.2): General conversation and writing
High (1.3-1.5): When you want very diverse vocabulary"""
```

```python
"""“In conversation, don’t repeat same topic—bring something new”
Prompt: Write about artificial intelligence.
Without penalties, model might say:
Artificial intelligence is transforming artificial intelligence systems.
Artificial intelligence is used in artificial intelligence research.
Presence penalty → “You already used this word once? Try something new.”
Penalty is same if the word get repeated one time or 10 times
When to use:
· Positive (0.1-1.0): Reduce overuse of common words, encourage varied vocabulary
· Zero (0.0): Natural language patterns
· Negative (-0.1 to -1.0): When repetition of key terms is important"""
```

```python
"""Frequency penalty → “You are using this word too many times. Enough !!!!.”
Penalty gets appended more based on the repetion of word. If a word get repeated
more times
then it gets more penalty added.
When to use:
· Positive (0.1-1.0): Reduce overuse of common words, encourage varied vocabulary
· Zero (0.0): Natural language patterns
· Negative (-0.1 to -1.0): When repetition of key terms is important"""
```

```python
import openai
openai_client=openai
prompt = "Write a short poem about the moon and the ocean."
response = openai_client.chat.completions.create(
 model="gpt-4",
 messages=[
 {"role": "system", "content": "You are a creative poet."},
 {"role": "user", "content": prompt}
 ],
 temperature=1.2, # More creative and diverse
 top_p=0.9, # Nucleus sampling: use top 90% probable tokens
 presence_penalty=0.8, # Encourage new vocabulary
 frequency_penalty=0.5 # Reduce overused words
 # Note: repetition_penalty is not a parameter in the OpenAI API; you can omit it
)
print("\n--- Generated Output ---\n")
print(response.choices[0].message.content)
```
