```python
def call_llm(prompt,model: str = "gpt-4o"):
  from openai import OpenAI

  client = OpenAI()
  messages=[
        {
            "role":"user","content": prompt,
        },
    ]
  response = client.chat.completions.create (
        model=model, #Specify which model to use (e.g. gpt-4o,gpt-4)
        messages=messages, #Message History to provide context (only 1 user message here)
        temperature=0 #Control Randomness 0=Determibnistic, 1= Creative/Random

    )
  return response.choices[0].message.content


```

```python
prompt = f"""
What is the sentiment of the following product review,
which is delimited with triple backticks?
Review text: '''{lamp_review}'''
generate the output in the below JSON format:
{{
 "sentiment": "positive" or "negative or neutral",
 "reason": "reason for the sentiment"
}}
"""
response = call_llm(prompt)
print(response)
```
