```python
from openai import OpenAI

client = OpenAI()

completion = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {
            "role": "developer","content":"You are a helpful assistant to generate PythonCode"},
        {
            "role":"user","content": "Can you give the code for hello World?",
        },
    ],
)

print(completion.choices[0].message.content)
```

```python
def call_llm(prompt,model: str = "gpt-4o"):
  from openai import OpenAI

  client = OpenAI()

  completion = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {
            "role":"user","content": prompt,
        },
    ],
)

  print(completion.choices[0].message.content)
```

```python
call_llm("What is the usecase of AI in 2 lines")
```
