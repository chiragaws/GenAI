```python
from pinecone import Pinecone
from pinecone import ServerlessSpec

pc = Pinecone()

index_name = "chirag-vectors"

if not pc.has_index(index_name):
    pc.create_index(
        name=index_name,
        vector_type="dense",
        dimension=384,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
```

LISTING ALL THE INDEXES

```python
pc = Pinecone()
for index_model in pc.indexes.list():
  print(index_model.name,index_model.host,index_model.status.state)
```

DESCRIBING INDEX DETAILS

```python
index_model = pc.indexes.describe(name=index_name)
print(index_model.status, index_model.schema)

host = index_model.host
```

UPSERTING The RECORDS in INDEXES

```python
embeddings=model.encode(content).tolist()
index = pc.Index("chirag-vectors")

index.upsert(
    vectors = zip(ids,embeddings,metadata)
)
```

RETRIVING THE RECORDS in INDEXES

```python
query = "Coffee is very Strong"
```

```python
index = pc.Index("chirag-vectors")
search_embeddings= model.encode(query).tolist()
results=index.query(
    namespace="__default__",
    vector=search_embeddings,
    top_k=3,
    include_metadata=True,
    include_values=False
)
```

```python
results
```

```python
for document in results.matches:
  print(document.metadata["text"])
```
