````python
def call_llm(prompt,model: str = "gpt-4o"):
  from openai import OpenAI

  client = OpenAI()
  messages=[
        {
            "role":"user","content": prompt,
        },
    ]
  response = client.chat.completions.create (
        model=model, #Specify which model to use (e.g. gpt-4o,gpt-4)
        messages=messages, #Message History to provide context (only 1 user message here)
        temperature=0 #Control Randomness 0=Determibnistic, 1= Creative/Random

    )
  return response.choices[0].message.content


````

````python
"""Thousands of customers write reviews. No human can read all of them.
So systems summarize them into ‘Customers say’.”
What actually happens:
1. Collect all reviews
2. Process using ML/LLM
3. Identify patterns:
Positive
Negative
Neutral
4.Generate short summary
Example Output:
“Customers like the quality but mention the product is slightly overpriced."""
````

````python
prompt = f"""
Your task is to generate a short summary of a product \
review from an ecommerce site.
Summarize the review below, delimited by triple
backticks, in at most 30 words.
Review: ```{product_review}```
"""
response = call_llm(prompt)
print(response)
````
